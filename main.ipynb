```python
team_member = int(input("numbers of team members : "))
months = int(input("numbers of months: "))
monthly_salary = float (input("monthly salary : "))
cloud_cost = float (input("total cloud cost: "))

salary_cost =team_member * months * monthly_salary
subtotal = salary_cost + cloud_cost
vat = subtotal * 0.14
total = subtotal + vat

print(f"salary cost ={salary_cost:.2f}, subtotal = {subtotal:.2f}, vat= {vat:.2f}, total = {total:.2f}")
```

```python
message = input("Enter customer message: ")
hours = int(input("Enter hours open: "))

if message == "payment":
    team = "Billing"
elif message == "delivery":
    team = "Logistics"
elif message == "error":
    team = "Tech"
else:
    team = "General"

if hours >= 24:
    priority = "HIGH"
else:
    priority = "LOW"

print("Team:", team)
print("Priority:", priority)
```
